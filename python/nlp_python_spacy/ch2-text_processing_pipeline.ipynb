```python
import spacy

# Create an instance of the Language class
nlp = spacy.load('en_core_web_sm')
# Create a Doc object. A Doc object is a container for a sequence of Token objects.
doc = nlp("I am flying to Frisco")
print([w.text for w in doc])
```

```python
print(f'nlp is of type {type(nlp)}')
print(f'doc is of type {type(doc)}')
print(f'doc[2] ("{doc[2]}") is of type {type(doc[0])}')
```

```python
# Lemmatization: reducing a word to its base or dictionary form
doc = nlp(u'this product integrates both libraries for downloading and applying patches')
for token in doc:
    print(token.text, token.lemma_)
```

```python
patterns = [
    [{"TEXT": "Frisco"}],  # Exact match
    [{"TEXT": "frisco"}],  # Lowercase variant
    [{"TEXT": "FRISCO"}],  # Uppercase variant
]

ruler = nlp.get_pipe("attribute_ruler")
ruler.add(patterns=patterns, attrs={"LEMMA": "San Francisco"})
doc = nlp(u"I am flying to Frisco")
for token in doc:
    print(token.text, token.lemma_)
```

```python
# Part of speech (POS) tagging: assigning a part of speech to each token
doc = nlp(u'I have flown to LA. Now I am flying to Frisco.')
print([w.text for w in doc if w.tag_ == 'VBG' or w.tag_ == 'VB'])  # VBG is the tag for gerunds
print([w.text for w in doc if w.pos_ == "PROPN"]) # PROPN is the tag for proper nouns
```

```python
# .pos_ is the part of speech
# .dep_ is the syntactic dependency relation

for token in doc:
    print(token.text, token.pos_, token.dep_)
```

```python
# The head property refers to the syntactic head of the token, which is the word that the token is dependent on.
# The dep_ property refers to the dependency relation between the token and its head.

for token in doc:
    print(token.head.text, token.head.head.text, token.dep_, token.text)
```

```python
# Locate words that are the root of a sentence or the object of a preposition to better understand the user's intent.
# The key to locating the correct proper noun is recognizing that the head's head is the verb gerund rather than the verb past tense.
# It looks like one can navigate the dependency tree upwards by chaining the .head property.

print(
    [
        token.lemma_
        for token in doc
        if (token.tag_ == "VBG")
        or (token.dep_ == "pobj" and token.head.head.tag_ == "VBG")
    ]
)
```

```python
# NER

import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp(u"I have flown to LA. Now I am flying to Frisco")
for token in doc:
    if token.ent_type != 0:
        print(token.text, token.ent_type_)
```
