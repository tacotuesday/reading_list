```python
# spaCy has two basic components: Containers and Pipelines.
# Containers are data structures that hold the data and the models.
# Token, Span, and Doc are the three main containers in spaCy and represent tokens, a phrase or sentence, and a document respectively.

import spacy
nlp = spacy.load("en_core_web_sm")

doc = nlp(u"A severe storm hit the beach.")
print([doc[i].text for i in range(len(doc))]) 
```

```python
# Create a Doc object using its constructor
from spacy.tokens.doc import Doc
from spacy.vocab import Vocab

# The Vocab() object is a storage container that provides vocabulary and lexicon information.
doc = Doc(Vocab(), words = [u'Hi', u'there'])
doc
```

```python
# Iterating over a token's leftward syntactic children
doc = nlp(u'I want a green apple.')
print([w for w in doc[4].lefts])

# Can also use the `.children` property to get the children of a token
print([w for w in doc[4].children])
```

```python
# The Doc.sents container for sentence-level information

doc = nlp(u"A severe storm hit the beach. It started to rain.")
for sent in doc.sents:
    print([sent[i] for i in range(len(sent))])
```
